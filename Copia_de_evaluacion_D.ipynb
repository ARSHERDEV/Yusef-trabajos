```python

```

2. Calcula los **días** de viaje (división entera de horas entre `24`) e imprímelos.
3. ¿El viaje dura más de `10` horas? Imprime el booleano.
4. ¿La velocidad es múltiplo de `1000`? (usa `%`) Imprime el booleano.
5. Guarda las horas en `h`. ¿`h` es mayor a `13`? Imprime el resultado.

## Pregunta 5 — Operadores matemáticos, variables y booleanos (0.5 pts)

```python

ancho  = 30   # cm
alto   = 20   # cm
```

1. Calcula el **área de la base** (`largo * ancho`) e imprímela.
2. Calcula el **volumen** (`largo * ancho * alto`) e imprímelo.
3. Calcula el **peso estimado** si cada cm³ pesa `0.002` kg e imprímelo.
4. ¿El volumen supera `25000` cm³? Imprime el booleano.
5. ¿El `largo` es el doble del `alto`? Imprime el booleano.

````python
largo  = 45
ancho  = 30
alto   = 20

# Tu solución

#area de la base

area = (largo * ancho)
print(area)

#volumen

volumen = (largo * ancho * alto)
print(volumen)

#el peso estimado

pesoEstimado = (volumen * 0.002)
print(pesoEstimado)

#comparar volumen
print(volumen > 2500)

#el largo es el doble de alto?

print(largo == (alto *2))
````
