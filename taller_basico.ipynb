## Ejercicio 12 — Reto integrador
Construye un mini perfil de estudiante usando solo variables y un diccionario:

1. Define `grado = "10A"`, `materia = "Python"` y `nota_aprobatoria = 60`.
2. Crea un diccionario `estudiante` con las llaves: `nombre`, `nota1`, `nota2`, `nota3`.
3. Calcula el promedio: `(nota1 + nota2 + nota3) / 3` y guárdalo en `promedio`.
4. Determina si aprobó: `promedio >= nota_aprobatoria` y guárdalo en `aprobo`.
5. Determina si tuvo alto rendimiento: promedio >= 80 **y** las tres notas >= 70.
6. Imprime un resumen con todas las variables.

````python
# Tu solución aquí
grado = "10A"
materia = "Python"
nota_aprobatoria = 60


estudiante = {
    "nombre": "Carlos",
    "nota1": 75,
    "nota2": 88,
    "nota3": 62
}


promedio = (estudiante["nota1"] + estudiante["nota2"] + estudiante["nota3"]) / 3


aprobo = promedio >= nota_aprobatoria


alto_rendimiento = (promedio) >= 80 and \
                   (estudiante["nota1"] >= 70) and \
                   (estudiante["nota2"] >= 70) and \
                   (estudiante["nota3"] >= 70)


print(f"******* Resumen del Estudiante ********")
print(f"Grado: {grado}")
print(f"Materia: {materia}")
print(f"Nota Aprobatoria: {nota_aprobatoria}")
print(f"Nombre del Estudiante: {estudiante['nombre']}")
print(f"Notas: {estudiante['nota1']}, {estudiante['nota2']}, {estudiante['nota3']}")
print(f"Promedio: {promedio:.2f}")
print(f"Aprobado: {aprobo}")
print(f"Alto Rendimiento: {alto_rendimiento}")
print(f"******************************************")
````

````python
from random import randint
#datos
import pandas as pd

totalPedidos = []

productos = [
    {"productos": "Combo", "tiempo": 4},
    {"productos": "Boxes", "tiempo": 3},
    {"productos": "PopCorn", "tiempo": 6},
    {"productos": "Hamburguesa", "tiempo": 4}
]
empleados = [
    {"nombre": "Melany"},
    {"nombre": "Wanda"},
    {"nombre": "Cataleya"},
    {"nombre": "Yusef"},
    {"nombre": "Sara"},
]


pedidos = []
for i in range(300):
  producto = productos[randint(0, 3)]
  empleado = empleados[i % len(empleados)]["nombre"]
  pedidos.append({
      "empleado": empleado,
      "producto": producto["productos"],
      "tiempo": producto["tiempo"],
      "timeradd": producto["tiempo"] + randint(1, 10)
  })


df = pd.DataFrame(pedidos)
df
````

````python
mela = 0
wanda = 0
cataleya = 0
sara = 0
yusef = 0

total = {}
for i in pedidos:
  if i["empleado"] == "Melany":
     mela += i["timeradd"]
  if i["empleado"] == "Wanda":
     wanda += i["timeradd"]
  if i["empleado"] == "Yusef":
     yusef += i["timeradd"]
  if i["empleado"] == "Cataleya":
     cataleya += i["timeradd"]

  if i["empleado"] == "Sara":
     sara += i["timeradd"]

  total["Wanda"] = wanda / 60
  total["Melany"] = mela / 60
  total["Cataleya"] = cataleya / 60
  total["yusef"] = yusef / 60

  total["sara"] = sara / 60
print(total)

# Obtener la llave con el valor mínimo
empleado_min = min(total, key=total.get)
valor_min = total[empleado_min]

print(f"El más bajo es {empleado_min} con {valor_min:.2f} es la nueva feje")

````
